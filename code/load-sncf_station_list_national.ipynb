```python
import pandas as pd
import numpy as np
from pyprojroot import here
```

```python
DATA_RAW = here("data/raw")
DATA_PROCESSED = here("data/processed")
```

```python
# Load SNCF station list
stations_raw = pd.read_csv(
    DATA_RAW / "sncf_station_list_national.csv",
    sep=";",
    dtype=str
)

# 1. Basic structure
print("Shape:", stations_raw.shape)
print("Columns:", stations_raw.columns.tolist())

# 2. NA check
print("\nNA counts:")
print(stations_raw.isna().sum())

# 3. Key sanity checks
print("\nUnique CODE_UIC:", stations_raw["CODE_UIC"].nunique())
print("Unique LIBELLE:", stations_raw["LIBELLE"].nunique())

#Code changes
yn_map = {"O": "Yes", "N": "No"}
stations_raw["freight_service"] = stations_raw["FRET"].map(yn_map).fillna(stations_raw["FRET"])
stations_raw["passenger_service"] = stations_raw["VOYAGEURS"].map(yn_map).fillna(stations_raw["VOYAGEURS"])

#fixing feature names
stations_clean = stations_raw.rename(columns={
    "CODE_UIC": "uic",
    "LIBELLE": "name",
    "CODE_LIGNE": "line_code",
    "COMMUNE": "commune",
    "DEPARTEMEN": "department",
})[[
    "uic",
    "name",
    "line_code",
    "RG_TRONCON",
    "PK",
    "freight_service",
    "passenger_service",
    "commune",
    "department",
    "X_L93",
    "Y_L93",
    "X_WGS84",
    "Y_WGS84",
    "C_GEO",
    "Geo Point",
    "Geo Shape"
]]

# Integrity: uic present on all rows
assert stations_clean["uic"].notna().all()

stations_clean.to_csv(DATA_PROCESSED / "sncf_station_list_national-p.csv", index=False)

stations_clean.head()
```
